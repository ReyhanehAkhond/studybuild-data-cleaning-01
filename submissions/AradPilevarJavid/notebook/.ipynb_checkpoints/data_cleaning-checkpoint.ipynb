```python
import numpy as np
import pandas as pd
```

```python
path = "../First_Dataset.xlsx"
```

```python
df = pd.read_excel(path)
```

first I will try to understand the data frame

```python
df.describe()
```

```python
df.head()
```

We can already see that the genders aren't written correctly so we have to fix that

print names without indices

```python
for name in df["first_name"]:
    print(name)
```

based on the above results I understood that the diversity of the male names aren't much so I used a list of all of them.

```python
male_names = {
    "Reza",
    "Sina",
    "Parsa",
    "Amir",
    "Arash",
    "Ali",
}
```

```python
df['gender'] = df['first_name'].apply(
    lambda name: "M" if name in male_names else "F"
)
```

as we can see the genders are fixed. now It's time to remove the duplicates

```python
df = df.drop_duplicates()
```

```python
df
```

```python
# the spelling of Guilan was wrong
df["province"] = df["province"].replace("Gilan", "Guilan")
```

as the median is 45 the person who entered 145 as their age might also be 45:

```python
df.loc[9, 'age'] = 45
```

now I would try to understand if there is any bug in the rest of the columns

Consider the discounts and returned items.

```python
for row in range(len(df)):
    if (df.loc[row, "purchase_count"] >= df.loc[row, "returned_items"]):
        pass
    else:
        print(f"Row {row} must be fixed")
```

```python
l = [7, 14, 23, 28, 36, 55]
print(df.loc[l, ["avg_order_value","purchase_count", "returned_items", "satisfaction_score"]])
```

```python
df.loc[7, "purchase_count"] = 8
df.loc[14, "returned_items"] = 3
df.loc[23, "purchase_count"] = 2
df.loc[28, "returned_items"] = 2
df.loc[36, "purchase_count"] = 7
df.loc[55, "purchase_count"] = 4
```

```python
for row in range(len(df)):
    if (df.loc[row, "avg_order_value"] * df.loc[row, "purchase_count"] >= df.loc[row, "total_spending"]):
        pass
    else:
        print(f"Row {row} must be fixed")
```

```python
l = [4, 29, 35, 39, 52]

expected_total_spending = (
    df.loc[l, "avg_order_value"] * df.loc[l, "purchase_count"]
)

print(expected_total_spending)

print(df.loc[l, ["avg_order_value","purchase_count", "total_spending"]])
```

rows 29 and 39 must be fixed

```python
df.loc[39, "total_spending"] = 1280.44
df.loc[29, "total_spending"] = 4079.14
```
