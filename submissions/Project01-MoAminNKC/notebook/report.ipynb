```python
import pandas as pd
from ydata_profiling import ProfileReport
```

```python
# Load your perfectly clean dataset
df = pd.read_excel("../data/cleaned_dataset.xlsx")

# Rename the columns to clean, readable titles
# You pass a dictionary where the key is the old name and the value is the new name
df = df.rename(columns={
    "customer_id": "Customer ID",
    "first_name": "First Name",
    "gender": "Gender",
    "age": "Age",
    "city": "City",
    "province": "Province",
    "signup_date": "Signup Date",
    "membership_tier": "Membership Tier",
    "purchase_count": "Purchase Count",
    "avg_order_value": "Average Order Value",
    "total_spending": "Total Spending",
    "last_purchase_days": "Days Since Last Purchase",
    "payment_method": "Payment Method",
    "device": "Device",
    "discount_used": "Discount Used",
    "returned_items": "Returned Items",
    "satisfaction_score": "Satisfaction Score"
})

# Generate the report with the new, pretty column names
profile = ProfileReport(df, title="Customer Data Quality & EDA Report", explorative=True)
profile.to_file("../report/customer_analysis_report.html")
```
